## Step 4 — Read an image from a web address

A **URL** is a web address.
This example reads a small PNG image from Python's official website.

**Do:** Run the next cell.
This step needs an internet connection.
It should print the number of bytes received.

`urlopen` opens the address.
`response.read()` reads the image into `image_bytes`.
The `with` block closes the connection when it finishes.

If you get a network error, stop here and share the error message.


```python
from urllib.request import urlopen

image_url = "https://www.python.org/static/community_logos/python-logo.png"

with urlopen(image_url, timeout=15) as response:
    image_bytes = response.read()

print(f"Received {len(image_bytes):,} bytes.")
```

## Step 5 — Display text and an image

**Do:** Run the next cell.

`Markdown` formats the greeting as bold text.
`Image` tells Jupyter how to display the bytes.

You should see a greeting and the Python logo.


```python
from IPython.display import Image, Markdown, display

display(Markdown(f"**{greeting}**"))
display(Image(data=image_bytes, width=image_width))
```

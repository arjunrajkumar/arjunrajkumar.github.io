On this testnet address `n2UXqRMWykSiXUqAtLXr9UmNBTZUnsq5is`, and index 0, I have 0.00165654 bitcoins.

First, i'm going to create a new transaction to send 0.005 bitcoins to a new address, and 0.005 back to me. So, we have one input and two outputs.

```python
import sys
from pathlib import Path

chapter_dir = str(Path("../programmingbitcoin/book/code-ch07").resolve())
if chapter_dir not in sys.path:
    sys.path.insert(0, chapter_dir)

from ecc import PrivateKey
from helper import decode_base58, SIGHASH_ALL, hash256, little_endian_to_int
from script import p2pkh_script, Script
from tx import TxIn, TxOut, Tx
from getpass import getpass
```

```python
target_address = 'mwJn1YPMq7y5F8J3LkC5Hxg9PHyZ5K4cFv'
prev_tx_id = '75650387d72fe18103cb8871c4df9e5d1472ec8f44ebecb60a712f2fe1e72fce'
my_public_testnet_address = 'n2UXqRMWykSiXUqAtLXr9UmNBTZUnsq5is'

# creating the transaction
prev_tx = bytes.fromhex(prev_tx_id)
prev_index = 0
tx_ins = []
tx_ins.append(TxIn(prev_tx, prev_index))

tx_outs = []
target_amount = int(50000)
target_h160 = decode_base58(target_address)
target_script = p2pkh_script(target_h160)
tx_outs.append(TxOut(amount=target_amount, script_pubkey=target_script))

change_amount = int(90000)
change_h160 = decode_base58(my_public_testnet_address)
change_script = p2pkh_script(change_h160)
tx_outs.append(TxOut(amount=change_amount, script_pubkey=change_script))

tx_obj = Tx(1, tx_ins, tx_outs, 0, testnet=True)

# signing the transaction
passphrase = getpass("Chapter 4 testnet passphrase: ").encode("utf-8")
secret = little_endian_to_int(hash256(passphrase))
private_key = PrivateKey(secret)
print(tx_obj.sign_input(0, private_key))
```

Next, lets create a new transaction to bitcoins to a new address from the above address, and also from a new input faucet address to the same target address, and some change back to me. So, we have two inputs and two outputs.

```python
new_prev_tx_id = '5b2630e6c8d870c2f94246e6229f93dedefed0ef8223e17d755a4964baa26b6e'

# creating the transaction
# creating first input from old transaction
prev_tx = bytes.fromhex(new_prev_tx_id)
prev_index = 1
tx_ins = []
tx_ins.append(TxIn(prev_tx, prev_index))
# creating second input from new faucet address
prev_tx_id_for_new_faucet = '25cf43ce915af30ccca2e0e30541d25a403c43d4c8b3d6d621757ed7d35c73ed'
public_testnet_address_for_new_faucet = 'n37oxw75WksUhBY56VPsSHzPrS4bKfJyUv'
prev_tx = bytes.fromhex(prev_tx_id_for_new_faucet)
prev_index = 1
tx_ins.append(TxIn(prev_tx, prev_index))

tx_outs = []
target_amount = int(20000)
target_h160 = decode_base58(target_address)
target_script = p2pkh_script(target_h160)
tx_outs.append(TxOut(amount=target_amount, script_pubkey=target_script))

change_amount = int(220000)
change_h160 = decode_base58(my_public_testnet_address)
change_script = p2pkh_script(change_h160)
tx_outs.append(TxOut(amount=change_amount, script_pubkey=change_script))

tx_obj = Tx(1, tx_ins, tx_outs, 0, testnet=True)

# signing the two inputs
def get_signing_key(prompt, expected_address):
    passphrase = getpass(prompt).encode("utf-8")
    secret = little_endian_to_int(hash256(passphrase))
    key = PrivateKey(secret)
    if key.point.address(testnet=True) != expected_address:
        raise ValueError("This passphrase gives a different testnet address.")
    return key

old_key = get_signing_key("Old address passphrase: ", my_public_testnet_address)
new_key = get_signing_key("New address passphrase: ", public_testnet_address_for_new_faucet)

print("Old input:", tx_obj.sign_input(0, old_key))
print("New input:", tx_obj.sign_input(1, new_key))
```

Came across this old question on stack overflow where the user wants to learn how to uncompress a public key. This is important as Bitcoin serialises public keys (or transactions, digital signatures) before transmitting them to other computers on the network.

The user is asking how to uncompress a public key from compressed public keys. First, lets look at what is an uncompressed public key.

##### Uncompressed public keys
A public key in Bitcoin is basically (x,y) coordinates, and an uncompressed public key looks like this:

`04ad90e5b6bc86b3ec7fac2c5fbda7423fc8ef0d58df594c773fa05e2c281b2bfe877677c668bd13603944e34f4818ee03cadd81a88542b8b4d5431264180e2c28`

First byte is the prefix byte (04); Next 32 bytes is the x coordinates as a big-endian integer; Next 32 bytes is the y coordinates as a big-endian integer.

This is the secp256k1 elliptic curve equation: `y² = x³ + ax + b `

Here, both (x,y) and (x,-y) will solve this equation, so for every x coordinate, there are at most 2 y coordinates.

##### Compressed public key
To convert a public key into a the compressed key we first start with the prefix byte. If y is even, it’s 0x02; otherwise, it’s 0x03.
In the above key, these are the x,y coordinates:
```
x: ad90e5b6bc86b3ec7fac2c5fbda7423fc8ef0d58df594c773fa05e2c281b2bfe
y: 877677c668bd13603944e34f4818ee03cadd81a88542b8b4d5431264180e2c28
```
We can see that the y-coordinates ends in 2c28. The final digit is 8, so y is even and the compressed key starts with 02.

So, the compressed public key is:
`0x02ad90e5b6bc86b3ec7fac2c5fbda7423fc8ef0d58df594c773fa05e2c281b2bfe`

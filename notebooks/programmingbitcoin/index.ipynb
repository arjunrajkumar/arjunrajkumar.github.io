### Exercise 1.7

<span id="work-1-7" data-learnt="2026-11-04" data-topic="Exercise 1.7 — Powers modulo a prime"></span>For p = 7, 11, 17, 31, what is this set in $F_{p}$?

{$1^{(p-1)}$, $2^{(p-1)}$, $3^{(p-1)}$, $4^{(p-1)}$, ... $(p-1)^{(p-1)}$}

<details>
<summary>Book starter code and hints</summary>

```python
primes = [7, 11, 17, 31, 43]
```

</details>

````python
for prime in [7, 11, 17, 31]:
    exponent = prime - 1
    values = []
    for n in range(1, prime):
        values.append(pow(n, exponent, prime))
    print(prime, values)
````

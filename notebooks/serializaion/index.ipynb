Came across this old question on stack overflow, and tried to solve it. Basically, the user wants to learn how to uncompress a public key. This is important as Bitcoin serialises public keys (or transactions, digital signatures) before transmitting them to other computers on the network.

#### Uncompressed public keys
A public key in Bitcoin is basically (x,y) coordinates, and an uncompressed public key looks like this:

`04ad90e5b6bc86b3ec7fac2c5fbda7423fc8ef0d58df594c773fa05e2c281b2bfe877677c668bd13603944e34f4818ee03cadd81a88542b8b4d5431264180e2c28`

First byte is the prefix byte (04); Next 32 bytes is the x coordinates as a big-endian integer; Next 32 bytes:s the y coordinates as a big-endian integer

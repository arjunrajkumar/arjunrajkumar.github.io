<span id="division" data-learnt="2026-11-04" data-topic="Division in a finite field"></span>In finitie fields of range $p$ ($\mathbb{F}_p$), for two numbers $a$ & $b$,\
$\frac{a}{b}$ .. is same as writing $a \cdot b^{-1}$

$b^{-1}$ can be written as $b^{-1} \cdot 1$\
Because of multiplicative identitiy

$b^{-1} \cdot 1$  can be written as $b^{-1} \cdot b^{p-1}$\
Because we know that $b^{p-1} = 1$ always

$b^{-1} \cdot b^{p-1}$  can be written as $b^{p-2}$

so, $\frac{a}{b}$ is same as writing $a \cdot b^{p-2}$

If $p = 19$, then $\frac{a}{b}$ is same as writing $a \cdot b^{17}$\
Because $b^{-1}$ is same as $b^{17}$ for any value of $b > 0$.

So, in other words.. we can easily calculate in $\mathbb{F}_{19}$:\
$\frac{4}{8} = 4 \cdot 8^{17}$\
$\frac{5}{9} = 5 \cdot 9^{17}$

Division is an expensive operation as exponentials can grow fast. Hence in finite fields we also use the modulo operator to make the division faster. In Python, the pow function has an optional third argument. For e.g. in $4 \cdot 8^{17}$, we can calculate $8^{17}$ modulo $19$ with `pow(8, 17, 19)`. This gives the same result as `(8 ** 17) % 19` and avoids calculating the full power first.
